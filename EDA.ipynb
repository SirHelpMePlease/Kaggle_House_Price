```python
numericalcols = data_train.select_dtypes(include="number").columns
categorical_cols = data_train.select_dtypes(include=["object", "category"]).columns
```

```python
data_train.corr(numeric_only=True)["SalePrice"].sort_values(ascending=False)
```

```python
grand_mean = data_train["SalePrice"].mean()
ss_total = ((data_train["SalePrice"] - grand_mean) ** 2).sum()

eta_dict = {}

# Wir wandeln die Spalten hier direkt in eine Liste von Strings um
for col in categorical_cols.tolist():
    # Jetzt wird sauber nach einer einzelnen Spalte (String) gruppiert
    groups = data_train.groupby(col)["SalePrice"]

    ss_between = sum(len(g) * (g.mean() - grand_mean)**2 for _, g in groups)
    eta_dict[col] = ss_between / ss_total if ss_total != 0 else 0

eta_scores = pd.Series(eta_dict).sort_values(ascending=False)
print(eta_scores)
```

```python
skew_vals = data_train[numericalcols].skew().sort_values(ascending=False)
```
